````java
import java.util.*;
````

## 1. Define the Binary Tree Node structure

````java
class TreeNode {
    int val;
    TreeNode left;
    TreeNode right;
    TreeNode(int val) { this.val = val; }
}
````

## 2. Implement an Algorithm (Example: Inorder Traversal - Left, Root, Right)

````java
public List<Integer> inorderTraversal(TreeNode root) {
    List<Integer> result = new ArrayList<>();
    traverse(root, result);
    return result;
}
````

````java
private void traverse(TreeNode node, List<Integer> result) {
    if (node == null) return;
    traverse(node.left, result);   // Go Left
    result.add(node.val);          // Visit Root
    traverse(node.right, result);  // Go Right
}
````

## 3. Test and run the implementation

````java
TreeNode root = new TreeNode(1);
root.right = new TreeNode(2);
root.right.left = new TreeNode(3);

List<Integer> output = inorderTraversal(root);
System.out.println("🌳 Inorder Traversal Result: " + output);
// Expected Output: [1, 3, 2]
````
