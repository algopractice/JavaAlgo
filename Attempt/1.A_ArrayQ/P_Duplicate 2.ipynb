# Duplicate within K dist

```java
class WSolution {
    public boolean containsNearbyDuplicate(int[] nums, int k) {
        Map<Integer,Integer> fm = new HashMap<>();

        for(int i =0; i < nums.length; i++){

            if(fm.containsKey(nums[i])){
                int j = fm.get(nums[i]);
                if(i-j <= k){
                    return true;
                }
            }else{
                fm.put(nums[i],i);
            }
        }
        return false;
    }
}
```

# This is Wrong

We need to put in map in both the condition, whether key exists or not. 
- If key does not exist, put it in map, as usual
- If it exist, we still **need to update the index to latest**

```java
import java.util.*;

public class Main {
    public static void main(String[] args) {
        WSolution sol = new WSolution();

        // Test Case 1: Simple duplicate within range
        // Indices 0 and 3, distance 3 <= 3 (True)
        int[] nums1 = {1, 2, 3, 1};
        int k1 = 3;
        System.out.println("Test 1 (Expected: true): " + sol.containsNearbyDuplicate(nums1, k1));

        // Test Case 2: Duplicates exist, but distance is greater than k
        // Indices 0 and 3, distance 3 > 1 (False)
        int[] nums2 = {1, 2, 3, 1};
        int k2 = 1;
        System.out.println("Test 2 (Expected: false): " + sol.containsNearbyDuplicate(nums2, k2));

        // Test Case 3: Multiple duplicates, one pair satisfies condition
        // Indices 2 and 3 are distance 1 <= 1 (True)
        int[] nums3 = {1, 0, 1, 1};
        int k3 = 1;
        System.out.println("Test 3 (Expected: true): " + sol.containsNearbyDuplicate(nums3, k3));

        // Test Case 4: No duplicates
        int[] nums4 = {1, 2, 3, 4, 5};
        int k4 = 2;
        System.out.println("Test 4 (Expected: false): " + sol.containsNearbyDuplicate(nums4, k4));

        // Test Case 5: Empty array
        int[] nums5 = {};
        int k5 = 0;
        System.out.println("Test 5 (Expected: false): " + sol.containsNearbyDuplicate(nums5, k5));
        
        // Test Case 6: Duplicate at the very start and end
        // Distance is 4, k is 3 (False)
        int[] nums6 = {1, 2, 3, 4, 1};
        int k6 = 3;
        System.out.println("Test 6 (Expected: false): " + sol.containsNearbyDuplicate(nums6, k6));
    }
}
```
