# Given an array arr[]. Your task is to find the minimum and maximum elements in the array.

# Hidden Edge Case
If the input array is empty (arr.length == 0) or null, without the guard cluse your code will bypass the loop entirely and return [2147483647, -2147483648] (the raw sentinel values) or NPE, which can cause downstream logic failures.

```java
import java.util.ArrayList;
import java.util.List;

class Solution {

    public static ArrayList<Integer> getMinMax(int[] arr) {
        ArrayList<Integer> res = new ArrayList<>();
        
        // Guard clause for empty or null arrays
        if (arr == null || arr.length == 0) {
            return res; 
        }

        int max = arr[0];
        int min = arr[0];

        // Single-pass linear scan starting from index 1
        for (int i = 1; i < arr.length; i++) {
            if (arr[i] > max) {
                max = arr[i];
            } 
            if (arr[i] < min) {
                min = arr[i];
            }
        }

        res.add(min);
        res.add(max);
        return res;
    }
}
```
