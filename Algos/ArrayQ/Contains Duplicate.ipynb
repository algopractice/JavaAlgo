# Contains Duplicate

Given an integer array nums, return true if any value appears at least twice in the array, and return false if every element is distinct.
[Leetcode](https://leetcode.com/problems/contains-duplicate/submissions/2073884019/)

```java
class Solution {

    public boolean optimalSol(int[] nums) {
        Set<Integer> seen = new HashSet<>();
        for (int num : nums) {
            if (!seen.add(num)) {
                return true;
                }
            }
        return false;
        }
    
    public boolean sol1(int[] nums) {
        Map<Integer,Integer> fm = new HashMap<>();

        for(int num : nums){
            fm.put(num,fm.getOrDefault(num,0)+1);
        }

        for(Map.Entry<Integer,Integer> e : fm.entrySet()){
            if(e.getValue()>1){
                return true;
            }
        }
        return false;
    }
}
```

```java
public class Main {
    public static void main(String[] args) {
        Solution sol = new Solution();

        // Test Case 1: Standard case with a duplicate
        int[] nums1 = {1, 2, 3, 1};
        System.out.println("Test 1 (Expected: true): " + sol.optimalSol(nums1));

        // Test Case 2: No duplicates
        int[] nums2 = {1, 2, 3, 4};
        System.out.println("Test 2 (Expected: false): " + sol.optimalSol(nums2));

        // Test Case 3: Empty array
        int[] nums3 = {};
        System.out.println("Test 3 (Expected: false): " + sol.optimalSol(nums3));

        // Test Case 4: Single element array
        int[] nums4 = {1};
        System.out.println("Test 4 (Expected: false): " + sol.optimalSol(nums4));

        // Test Case 5: Duplicate at the very end
        int[] nums5 = {1, 2, 3, 4, 1};
        System.out.println("Test 5 (Expected: true): " + sol.optimalSol(nums5));

        // Test Case 6: All elements same
        int[] nums6 = {5, 5, 5, 5};
        System.out.println("Test 6 (Expected: true): " + sol.optimalSol(nums6));


        //---
        
        // Test Case 1: Standard case with a duplicate
        int[] nums11 = {1, 2, 3, 1};
        System.out.println("Test 1 (Expected: true): " + sol.sol1(nums1));

        // Test Case 2: No duplicates
        int[] nums12 = {1, 2, 3, 4};
        System.out.println("Test 2 (Expected: false): " + sol.sol1(nums2));

        // Test Case 3: Empty array
        int[] nums13 = {};
        System.out.println("Test 3 (Expected: false): " + sol.sol1(nums3));

        // Test Case 4: Single element array
        int[] nums14 = {1};
        System.out.println("Test 4 (Expected: false): " + sol.sol1(nums4));

        // Test Case 5: Duplicate at the very end
        int[] nums15 = {1, 2, 3, 4, 1};
        System.out.println("Test 5 (Expected: true): " + sol.sol1(nums5));

        // Test Case 6: All elements same
        int[] nums16 = {5, 5, 5, 5};
        System.out.println("Test 6 (Expected: true): " + sol.sol1(nums6));
    }
}
```
