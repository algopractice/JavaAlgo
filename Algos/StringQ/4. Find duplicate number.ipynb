```java
import java.util.*;
public int findDuplicate2(int[] nums) {
    Map<Integer, Integer> fr = new HashMap<>();
    int result = -1;
    for (int num : nums) {
        fr.put(num, fr.getOrDefault(num, 0) + 1);
    }
    for (Map.Entry<Integer, Integer> en : fr.entrySet()) {
        if (en.getValue() > 1) {
            result = en.getKey();
            break;
        }
    }
    return result;
}
```
