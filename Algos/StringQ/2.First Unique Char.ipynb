# Optimized Solution

 (from gemini)

* Use a simple integer array of size 26 (since LeetCode specifies the input consists of lowercase English letters).

```java
class Solution {
    public static int firstUniqChar(String s) {
        // Array to store frequencies of 'a' through 'z'
        int[] count = new int[26];
        
        // Step 1: Build the frequency map
        for (int i = 0; i < s.length(); i++) {
            count[s.charAt(i) - 'a']++;
        }
        
        // Step 2: Find the first character with a frequency of 1
        for (int i = 0; i < s.length(); i++) {
            if (count[s.charAt(i) - 'a'] == 1) {
                return i; // Return the index immediately
            }
        }
        
        return -1; // No unique character found
    }
}
```

## Bug :  
* If a unique character r is found, this loop keeps running until the end of the string.
* If r happens to be \0 (meaning no unique character was found), and your string actually contains a null character, it could cause issues.
* More importantly, because it doesn't break once it finds the character, it wastes time spinning through the rest of the string.

While adding a break; inside that if statement fixes the bug, we can actually make the whole algorithm much faster and cleaner

## The Optimized Approach :
Instead of using a heavy LinkedHashMap and running three separate loops, you can use a simple integer array of size 26 (since LeetCode specifies the input consists of lowercase English letters).
* First Pass: Count the frequencies of each character using the array.
* Second Pass: Walk through the string s from left to right. The first character you hit that has a count of 1 is your winner. Return its index immediately.

This slims your time complexity down and drastically reduces memory overhead.
