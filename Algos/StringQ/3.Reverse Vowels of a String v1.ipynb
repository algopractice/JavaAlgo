# Optimized Implementation 1

````java
public String reverseVowels(String s) {
    // 1. Define the search space (including uppercase)
    Set<Character> vowels = Set.of('a', 'e', 'i', 'o', 'u', 'A', 'I', 'E', 'O', 'U');

    int i = 0;
    int j = s.length() - 1;
    
    // 2. Convert to mutable char array for performance and direct modification
    char[] cArr = s.toCharArray();

    while (i < j) {
        // Phase 1: Left pointer is not a vowel
        if (!vowels.contains(cArr[i])) {
            i++;
        }
        // Phase 2: Right pointer is not a vowel
        else if (!vowels.contains(cArr[j])) {
            j--;
        }
        // Phase 3: Both pointers found a vowel -> SWAP!
        else {
            char temp = cArr[i];
            cArr[i] = cArr[j];
            cArr[j] = temp;
            
            // Advance both to avoid getting stuck
            i++;
            j--;
        }
    }
    // 3. Convert the modified primitive array back into a Java String
    return new String(cArr);
}
````

# Problem: Reverse Vowels of a String
### Technique: Two-Pointer Pattern (Moving Inward)

## Initial Mental Model & Syntax Hurdles
We started with the intuition of using two pointers (`i` at the start, `j` at the end) moving toward each other to find and swap vowels.

### Key Java Syntax Fixes:
* **Set Initialization:** Maps use `.put()`, but `Set` uses `.add()`. To initialize a collection quickly with values, use `Set.of('a', 'e', ...);`
* <mark class="note">Note</mark> **String Immutability:** In Java, strings cannot be modified in place. Expressions like `s.charAt(i) = 'x'` will throw a compilation error. To manipulate characters, we must convert the string to a mutable structure like a `char[]` array using `s.toCharArray()`.

### The Core Logic: Conditional vs. Unconditional Movement
* **Unconditional Movement:** Used when reversing a whole string. Both pointers move every turn.
* **Conditional Movement:** Used when searching for specific targets (vowels). A pointer must **freeze** when it finds its target and wait for the other pointer to find its target.

### The Infinite Loop Trap
If both pointers look for non-vowels using independent `if` statements:
```java
if (!vowels.contains(cArr[i])) { i++; }
if (!vowels.contains(cArr[j])) { j--; }
```
When both pointers land on a vowel simultaneously, both conditions evaluate to false. Neither pointer moves, resulting in an infinite loop.

### The 3-Phase Solution (Using if-else if-else)
To fix the freeze, we structure the loop so that exactly one of three phases happens per iteration:

1. Phase 1 (if): Left pointer is on a consonant. Move i++.
2. Phase 2 (else if): Right pointer is on a consonant. Move j--.
3. Phase 3 (else): Both pointers are on vowels! Swap them, then manually move both i++ and j-- to step away.

### Performance Insight: Why cArr[i] over s.charAt(i)?
* Inside the loop, checking cArr[i] is significantly faster than calling s.charAt(i).
* s.charAt(i) is a method call that executes hidden index out-of-bounds checks every single time it runs.
* <mark class="note">Note</mark> cArr[i] is a direct primitive array lookup. It accesses the raw memory address instantly, making the loop lightweight and highly performant.

# Solution 2

Classic **nested while-loop variation** of the inward two-pointer approach.

This code solves the _exact_ same problem, but its execution strategy inside the main loop is entirely different. Let's break down the two major shifts: the **Control Flow** change and the **Data Structure** change.

## Control Flow: The "Fast-Forward" vs. "One Step at a Time"

In our previous code, the outer `while (i < j)` loop acted like a single clock tick. On _every single tick_, it evaluated exactly one condition (`if`, `else if`, or `else`) and moved a pointer by a maximum of **one step**.

In this new code, you have inner `while` loops:

Java

```
while (i < j && !vowels.contains(chars[i])) { i++; }

```

Instead of stepping once and checking everything over again, this inner loop acts like a **fast-forward button**. It tells the `i` pointer: _"Keep sprinting forward until you smash face-first into a vowel. Do not stop, do not pass Go."_

#### Why this is structurally different:

-   **Fewer total iterations of the outer loop:** The outer loop only runs when a swap is actually about to happen or when the pointers cross.
    
-   **The "Ah-Ha!" detail:** Notice how the inner loops _also_ have to check `i < j`? If they didn't, a string with zero vowels (like `"crypt"`) would cause `i++` to sprint right past the end of the array and throw an `ArrayIndexOutOfBoundsException`.
    

### 2. The Set Implementation: `LinkedHashSet` vs. `Set.of`

In your new snippet, you used:

Java

```
Set<Character> vowels = new LinkedHashSet<>();

```

While this works perfectly fine for the problem, from a systems standpoint, it changes things under the hood:

-   **`LinkedHashSet` Overhead:** A `LinkedHashSet` maintains a doubly-linked list running through its elements to preserve their _insertion order_. Since we only care about checking `contains()`, we don't care what order the vowels were inserted in. Maintaining that linked list adds a tiny bit of unnecessary memory overhead.
    
-   **`Set.of()` (Our previous way):** This returns an unmodifiable, highly optimized, immutable Set. It takes up the absolute bare minimum space in memory and is faster for lookups of small, fixed groups of items like our 10 vowels.
    

### Which one is better?

Honestly, both run in $O(N)$ time complexity and both are great.

However, many engineers prefer the **inner while-loops** approach you just posted because the code's reading flow is incredibly linear:

1.  Find next left vowel.
    
2.  Find next right vowel.
    
3.  Swap 'em.
