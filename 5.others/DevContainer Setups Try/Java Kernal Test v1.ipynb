# To run Juypter notebook through vsc in browser 

``jupyter lab --ip=0.0.0.0 --port=8888 --no-browser --allow-root

Once you run this
* wait for the small notification to pop up in the bottom right corner of your screen, and click Open in Browser to launch JupyterLab.
* If you don't see the popup, look at the bottom toolbar of your editor and click the Ports tab next to your terminal.

  * You will see port 8888 listed there; hover over it and click the globe icon (Open in Browser) to open it manually.

## Shut down Jupyter Server

If it does not ask to off the server

`` pkill -9 -f jupyter
