# Style TODO and Note Globally

````python
from IPython.core.display import HTML

HTML("""
<style>
    /* Style any heading containing the word TODO */
    h1:has(mark.todo), h2:has(mark.todo), h3:has(mark.todo), h4:has(mark.todo) {
        color: red !important;
    }
    /* Style any heading containing the word Note */
    h1:has(mark.note), h2:has(mark.note), h3:has(mark.note), h4:has(mark.note) {
        color: green !important;
    }
</style>
""")
````

Once you run that code cell, you can write your headings in Markdown cells using an invisible <mark> tag like this:  

```## <mark class="todo"></mark>TODO: Fix this function```
