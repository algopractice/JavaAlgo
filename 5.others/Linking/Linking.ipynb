## See Method

```java
public static void see(String label, String url) {
    String htmlLink = String.format("<a href='%s' target='_blank'>@see: %s</a>", url, label);
    display(htmlLink, "text/html");
}
public static void main(String[] args) {
    System.out.println("Fetching framework references...");
    see("Optional Documentation", "https://docs.oracle.com/javase/8/docs/api/java/util/Optional.html");
}

// Execute the main method
main(new String[]{});
```

Or call the static method directly

```java
see("Inbuilt display method of JJava Kernel", "https://dflib.org/jjava/docs/1.x/#_display");
```

* Before the kernel displays anything, Java needs to construct the raw HTML string. String.format is a built-in Java utility that injects variables into a template.
* %s placeholders: These tell Java where to insert your variables.
* Order matters: The first %s is replaced by the url variable. The second %s is replaced by the label variable.
* target='_blank': This standard HTML attribute ensures that when you click the link in your notebook, it opens safely in a brand-new browser tab rather than navigating your current notebook page away.

#### Completely removed target='_blank' to bypass Jupyter's iframe security block

# Internal Text Link Tool, through CellId

## Through a id tag

### HyperLink

```java
public static void linkToNote(String label, String relativePath, String cellId) {
    String fullUrl = String.format("%s#%s", relativePath, cellId);
    String html = String.format("<a href='%s' style='color: #1a73e8; font-weight: bold; text-decoration: none;'>[%s]</a>", fullUrl, label);
    display(html, "text/html");
}
```

### Link on a button

```java
public static void buttonToNote(String label, String relativePath, String cellId) {
    String fullUrl = String.format("%s#%s", relativePath, cellId);
    String buttonHtml = String.format(
        "<a href='%s'>" +
        "<button style='background-color: #f1f1f1; border: 1px solid #ccc; padding: 2px 6px; border-radius: 3px; cursor: pointer; font-size: 11px;'>" +
        "%s</button></a>", 
        fullUrl, label
    );
    display(buttonHtml, "text/html");
}
```

Testing Both

```java
public static void main(String[] args) {
    System.out.print("Linking cell in another notebook through <id>");
    
    // Test link
    linkToNote("Hyperlink", "Linked Notebook.ipynb", "link-note"); 
    
    // Test button
    buttonToNote("Button Link", "Linked Notebook.ipynb", "button-note"); 
    
    System.out.println();
}

main(new String[]{});
```

```java
public static void patternButton2(String label, String relativePath, String cellId) {
    // 1. DEFINE THE MISSING SYMBOL: Convert spaces in the file path to URL-safe strings
    String encodedPath = relativePath.replace(" ", "%20");
    
    // 2. Combine the path and the Markdown heading ID anchor
    String fullUrl = String.format("%s#%s", encodedPath, cellId);
    
    // 3. Render the anchor tag styled identically to your original cyber-button
    String markdownHtml = String.format(
        "<a href='%s' style='text-decoration: none; display: inline-block;'>" +
        "<button style='background-color: #00e676; color: #003300; border: none; " +
        "padding: 6px 14px; border-radius: 50px; font-weight: 900; font-size: 11px; " +
        "letter-spacing: 0.5px; cursor: pointer; text-transform: uppercase; " +
        "box-shadow: 0 4px 10px rgba(0, 230, 118, 0.5);'>" + 
        "💡 PATTERN: %s</button></a>", fullUrl, label
    );
    
    // 4. Force Jupyter to process it natively via markdown string parsing
    display(markdownHtml, "text/markdown");
}
```
