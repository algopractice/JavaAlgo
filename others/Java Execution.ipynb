# Execution of Main method

## Main.main(new String[0]);

Q. Why Main.main(new String[0]) and not just Main.main(new String[])

In Java, when you declare a method that accepts an array—such as the `main` method's `String[] args`—the language requires you to provide the **size** or an **initializer list** if you are creating an instance of that array explicitly.

Here is why `new String[0]` is used instead of `new String[]`:

-   **Syntax Rules:** In Java, when you use the `new` keyword to instantiate an array, you **must** provide the dimension (the size) or an array initializer.
    
-   **The Error:** `new String[]` without a size or curly braces is a syntax error because the compiler does not know how much memory to allocate for the array.
    
-   **The Solution (`new String[0]`):** This creates an empty array of `String` objects with a length of zero. It perfectly satisfies the compiler's requirement for a specific array size while passing an "empty" argument list to the `main` method.
    
-   **Alternative Syntax:** You could also write `new String[] {}`, which is an array initializer with no elements. This is also valid and equivalent to `new String[0]`.
    

In short, Java needs to know exactly how many slots to create in the array memory, and `[0]` explicitly tells the compiler to create an array with zero slots.
