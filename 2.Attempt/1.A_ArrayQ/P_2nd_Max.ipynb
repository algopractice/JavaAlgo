# Second Largest Element in Array

````java
class Solution {
    public static int secondLargestElement(int[] nums) {
        int max1 = Integer.MIN_VALUE; // largest
        int max2 = Integer.MIN_VALUE; // second largest
        boolean hasSecond = false;

        for (int i = 0; i < nums.length; i++) {
            if (nums[i] > max1) {
                max2 = max1;
                max1 = nums[i];
            } else if (nums[i] > max2 && nums[i] != max1) {
                max2 = nums[i];
                hasSecond = true;
            }
        }
        return hasSecond ? max2 : -1;
    }
}
````

````java
import java.util.*;
public class Main{
    public static void main(String[] args) {
        Solution sol = new Solution();
        // Test Case 1: Standard random array
        int[] t1 = {4, 1, 9, 3, 7};
        System.out.println("Test 1 (Expected 7): " + sol.secondLargestElement(t1));

        // Test Case 2: Duplicates at the top (The trickiest edge case!)
        int[] t2 = {10, 10, 5};
        System.out.println("Test 2 (Expected 5): " + sol.secondLargestElement(t2));

        // Test Case 3: All elements identical (Should return -1)
        int[] t3 = {5, 5, 5};
        System.out.println("Test 3 (Expected -1): " + sol.secondLargestElement(t3));

        // Test Case 4: Negative numbers
        int[] t4 = {-10, -3, -5, -1};
        System.out.println("Test 4 (Expected -3): " + sol.secondLargestElement(t4));

        // Test Case 5: Single element (Should return -1)
        int[] t5 = {42};
        System.out.println("Test 5 (Expected -1): " + sol.secondLargestElement(t5));
    }
}
````

````java
Main.main(new String[]{});
````

# Core Pattern & Strategy - Demotion Model
- **Goal:** Find the second largest distinct element in a single pass ($O(n)$ time complexity).
- **State Trackers:** `largest`, `secondLargestElement`, and a boolean validation flag (`hasSecond`).

## Demote both current max and second max while comparing current number

When `nums[i] > largest`, push the old `largest` down to `secondLargest` before updating `largest`.
    

```
if((num[i] > max1){
    max 2 = max1;
    max1 = num[];

```

If above condition is false,  compare `nums[i] > max2` and update max2. 
```
else if(nums[i] > max2){
    max2 = nums[i];
```
#### Note :
This will work if there are no duplicated in the array. But if there are duplicates, it will two pitfalls discussed below

## Common Pitfalls & Corrections

### <font color="#ff0000">Pitfall 1 - Duplicate Values</font>

Arrays like `[10, 10, 5]` incorrectly set both variables to `10`, as second 10 would go to else if block and become true
- _Fix:_ Add an explicit inequality check in the `else if` condition: `nums[i] != largest`.
- Check dry run below

### <font color="#ff0000">Pitfall 2 - Missing Second Element</font>

Arrays with uniform values (e.g., `[5, 5, 5]`) fail to trigger `secondLargest` updates, leaving it at `Integer.MIN_VALUE`.
- _Fix:_ Introduce a `hasSecond` boolean flag inside the `else if` block to definitively prove a valid runner-up was found.

### Dry run for pitfall 1

### The Test Case: `[10, 10, 5]`

Imagine we run this array through our algorithm *without* the `&& nums[i] != largest` check.

Let's trace the variables:

* `largest` starts at `Integer.MIN_VALUE`
* `secondLargestElement` starts at `Integer.MIN_VALUE`

Now let's loop through the elements one by one:

1. **Iteration 1 (`nums[i] = 10`):**
* Is `10 > largest` (`Integer.MIN_VALUE`)? **Yes!**
* `secondLargestElement` becomes the old `largest` (`Integer.MIN_VALUE`).
* `largest` becomes `10`.
* *State after step 1:* `largest = 10`, `secondLargestElement = Integer.MIN_VALUE`.


2. **Iteration 2 (`nums[i] = 10`):**
* Is `10 > largest` (`10`)? **False.** (Skipped the first `if`)
* Let's look at the `else if` condition **without** our duplicate check: `if (nums[i] > secondLargestElement)` $\rightarrow$ `if (10 > Integer.MIN_VALUE)`.
* Is that true? **Yes!**
* So `secondLargestElement` gets overwritten and becomes `10`!
* *State after step 2:* `largest = 10`, `secondLargestElement = 10`.


3. **Iteration 3 (`nums[i] = 5`):**
* Is `5 > largest` (`10`)? **False.**
* Is `5 > secondLargestElement` (`10`)? **False.**
* Loop ends.



### The Result of the Bug

At the end of the loop, both `largest` and `secondLargestElement` are equal to `10`. Your final check (`largest == secondLargestElement`) would trigger and return `-1`.

However, in the array `[10, 10, 5]`, there *is* a valid second-largest **distinct** element, which is `5`!

### How the Fix Solves It (`&& nums[i] != largest`)

When the loop hits the second `10`, the `else if` checks:
`if (10 > secondLargestElement && 10 != largest)` $\rightarrow$ `if (10 > Integer.MIN_VALUE && 10 != 10)`.

Because `10 != 10` evaluates to **false**, the condition fails. The code refuses to let `secondLargestElement` become `10` because `10` is already the current champion (`largest`). It safely preserves the integrity of our distinct runner-up values!
