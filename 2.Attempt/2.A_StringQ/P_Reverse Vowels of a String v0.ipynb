## Does all two pointers have conditional movement?
No, not all of them, but it depends entirely on the problem's goal.

Unconditional Movement: If you are reversing a whole string (like swapping index 0 with the last index, then index 1 with the second-to-last), both pointers move every single turn. No questions asked.

Conditional Movement: If you are looking for specific targets (like vowels), a pointer should only move if it hasn't found its target yet. If a pointer finds a target, it needs to freeze and wait for the other pointer to find its target.

How to Structure the Reasoning
To fix this, think of the while loop as having three distinct phases on every turn:

Left Pointer Check: If i is pointing at a consonant, it's not interesting. Move i++.

Right Pointer Check: If j is pointing at a consonant, it's not interesting. Move j--.

The Target Phase: If neither of the above is true, it means both i and j are standing on vowels! This is your moment to swap them. And after you swap them, you must manually move both i++ and j-- so they don't get stuck on those same vowels on the next turn.

In Java, String objects cannot be changed once created. Methods like s.charAt(i) only read characters; you cannot use them to modify the string, and expressions like s.charAt(i) = ... will cause a compilation error.

How do Java developers bypass this?
When we need to manipulate characters inside a string (like swapping elements), we temporarily convert the string into a mutable structure. You have two excellent options here:

Convert to a char[] array: ```java

StringBuilder is a class specifically designed for modifying strings dynamically. It provides a .setCharAt(index, character) method.

# Why is cArr[i] better than s.charAt(i)?
It all comes down to what happens under the hood when Java reads data.

When you call s.charAt(i), Java doesn't just instantly grab the character. It has to execute a method call, which involves:

Validating that your index i isn't out of bounds (doing a hidden if check).

Fetching the internal value from the string's memory.

When you do s.charAt(i) inside a loop that runs thousands of times, you are forcing Java to run that full method and its internal safety checks over and over again.

By contrast, cArr[i] is a direct primitive array lookup. It is a single, lightning-fast hardware operation that directly reads the memory address. Since you already paid the tiny price to create cArr at the beginning of the method, reading from it directly makes your loop much lighter and faster.
