#### The Trap of Creating Custom Classes Too Quickly
* Lesson: Always ask yourself: "Can I solve this using standard primitive types or built-in Java collections first?"
#### The Danger of the else Return
* If the very first character in your map is a duplicate (like the 'l' in "lovelife"), your code enters the else block and immediately returns -1 on the very first iteration, completely ignoring the rest of the string!
* Lesson: Be incredibly careful with return statements inside loops. Only return early if you are 100% certain no further iteration matters.

#### Java Generics Don't Accept Primitives
* Lesson: Java Generics do not support primitive types (char, int, double). You must always use their Object wrappers (Character, Integer, Double).

#### It is mandatory to initialize local variables?
* Lesson: Yes.
    * In Java, class fields get default values (like null or 0),
    * but local variables inside a method do not.
* If you try to read a local variable later (like if(c == '')) without giving it a starting value, Java will throw a compile-time error: "Variable 'c' might not have been initialized".

#### The Empty Character Literal Myth
* if(c == '').
* Lesson: In Java, an empty string literal "" is valid, but ** an empty character literal '' is syntax error**.
* A char must always contain exactly one character.
* If you need a "blank" character value, you either use a placeholder like '\0' (the null character) or switch to the object wrapper Character and initialize it to null.

#### Watch Out for Copy-Paste Infinite Loops
* You accidentally incremented the string reference s++ instead of your index pointer i++. This creates an infinite loop because i stays 0 forever.

#### "Don't I need an else here, in case none of the chars are unique?"
You commented: // don't i need else here, in case none of the chars are unique ?

* Lesson: No, you don't!
    * If none of the characters are unique, the condition if (e.getValue() == 1) will never evaluate to true.
    * The loop will finish naturally,
    * and c will remain exactly what you initialized it to: null.
      
Then, your subsequent check if (c == null) return -1; perfectly catches it. This is a great clean-coding pattern.

#### Trusting the Map vs. Trusting the String
In Attempt 3, you find the correct character c using the map, and then you spin through the original string to find its index. This works, but it's computationally heavier because you are searching through the string all over again.

This friction is ultimately what led you to your final breakthrough: realizing that instead of mapping characters to counts and trying to track back to the index, you could just map the array positions directly!

Instead of using a heavy LinkedHashMap and running three separate loops, you can use a simple integer array of size 26 (since LeetCode specifies the input consists of lowercase English letters).
* First Pass: Count the frequencies of each character using the array.
* Second Pass: Walk through the string s from left to right. The first character you hit that has a count of 1 is your winner. Return its index immediately.

# The static Block Trick (JIT Warm-up) in Leetcode


```java
static {
    for (int i = 0; i < 300; i++)
        firstUniqChar("");
}
```

In Java, a static block runs exactly once when the class is first loaded into memory by the Java Virtual Machine (JVM), before LeetCode even starts running the actual test cases against your code.

By calling firstUniqChar("") 300 times with an empty string, the author is deliberately forcing the JVM’s Just-In-Time (JIT) Compiler to wake up.

When the JVM notices a method is being called repeatedly (a “hotspot”), it compiles that method from slow bytecode into native machine code and optimizes it heavily.

Because LeetCode runs all its test cases in a single JVM session, warming up the method before the official timer starts means that when LeetCode executes the actual test cases, it executes the pre-compiled native machine code instantly. This shaves off a few milliseconds of JVM startup/interpretation overhead, landing them a 0ms or 1ms execution time.
