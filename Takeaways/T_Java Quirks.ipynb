## For loop in Java

``` for(int i = 0; i < chars.length; i+2)```

In Java, C++, and C#, the third part of a for loop *requires an assignment statement* to change the counter variable. Writing i+2 simply calculates the number but throws it away, leaving i stuck at 0 forever.

The Solutions

Depending on what you want to achieve, you can fix this statement in a few ways:

-   **To skip every other element (increment by 2)**:  
    Change `i+2` to **`i += 2`** or **`i = i + 2`**.
    
    ``` for(int i = 0; i < chars.length; i += 2)```
        
-   **To iterate through every single element (standard loop)**:  
    Change `i+2` to **`i++`**.
        
    ```for(int i = 0; i < chars.length; i++)```
        
-   **If `chars` is a String instead of an Array**:  
    Add parentheses to make it a method call: **`chars.length()`**.
        
    ```for(int i = 0; i < chars.length(); i += 2)```
