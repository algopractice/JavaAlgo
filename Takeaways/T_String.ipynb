# Regex

## Always escape special regex characters (like `.`, `+`, `*`, `?`) when using `.split()`.

split("\\.")

-   The dot `.` is a special character in regex that means "match any character."
    
-   If you use `.split(".")`, you are telling Java to split the string at _every single character_, which results in an empty array or unexpected behavior.
    
-   **The Fix:** You must escape the dot with double backslashes: `"\\."` 

## Why use double backslashes (```\\```) and double quotes (i.e not single quote)?

In Java, the backslash `\` is an **escape character** inside strings. If you want to use a literal backslash, you have to type `\\`.

When you pass a string to a regex method (like `split()` or `replaceAll()`), Java passes that string to the Regex engine. The Regex engine _also_ has its own special characters.

-   **The Problem:** The dot `.` in regex means "any character." If you want to split by a **literal dot** (the period symbol), you need to escape it for the regex engine: `\.`.
    
-   **The Conflict:** To get that `\.` into the regex engine, you have to escape the backslash _again_ for Java's string compiler.
    
-   **The Result:** `\\.`
    
    -   The Java compiler sees `\\` and converts it to a single literal `\`.
        
    -   The regex engine then sees `\.` and knows to treat the dot as a literal character rather than "any character."


Java is a compiled language. It doesn't have a built-in "Regex" data type like Javascript. When you call a method like `split(regex)`, the method is defined to accept a `String` as an argument. Because the Regex engine is **software** running inside the Java environment, the only way to "send" the instructions to that software is to package them inside a `String` variable. Other methods requiring the same are `replaceAll` and `Pattern.compile`

## Metacharacters

If you want to treat them as normal characters (like a literal `.` or a literal `+`), you must escape them with `\\` in Java:

## Normal characters don't need escaping

If a character is not a metacharacter (like `a`, `B`, `5`, or `-`), you just type it normally.
-   To split by the letter "a", your regex is just `"a"`.
-   To split by a comma, your regex is just `","`.

## The Danger of the Single Slash (`\`)

In Java, a single slash inside double quotes (`" "`) is strictly reserved for **Java's built-in String formatting**.

Java only recognizes a very specific list of single-slash commands:

-   `\n` (New line)
    
-   `\t` (Tab)
    
-   `\"` (Quote inside a string)
    
-   `\\` (Literal backslash)
    

If you try to escape a normal letter with a single slash—for example, `"\m"` or `"\z"`—the Java compiler will throw a fit and give you an **"Illegal escape character" error**. Your code won't even compile.

## Character Classes and Boundary Matchers

- Escaping a normal letter does the opposite of what you think!
    - In regex, escaping a metacharacter (like `\\.`) turns it into a normal character.
    - However, escaping a _normal_ character often turns it into a **metacharacter**.

If you use double backslashes before certain normal letters, you accidentally trigger hidden Regex superpowers. These are called Character Classes and Boundary Matchers.

Here is the complete list of those "hidden Regex superpowers." Notice how 
**the lowercase letters usually mean is this thing**, while the **uppercase letters mean the exact opposite** ("is NOT this thing").

| Regex Concept | What it Matches | How to write it in Java |
| :--- | :--- | :--- |
| `\d` | **Any digit** (0-9) | `"\\d"` |
| `\D` | **Any non-digit** (letters, symbols, spaces) | `"\\D"` |
| `\s` | **Any whitespace** (space, tab, newline) | `"\\s"` |
| `\S` | **Any non-whitespace** (visible characters) | `"\\S"` |
| `\w` | **Any word character** (a-z, A-Z, 0-9, and `_`) | `"\\w"` |
| `\W` | **Any non-word character** (punctuation, spaces) | `"\\W"` |
| `\b` | **A word boundary** (the invisible edge of a word) | `"\\b"` |
| `\B` | **A non-word boundary** (inside a word) | `"\\B"` |
| `\A` | The absolute **beginning** of the input | `"\\A"` |
| `\z` | The absolute **end** of the input | `"\\z"` |

## Pro-Tip for Memorizing:
-   **`d`** is for **D**igit.
-   **`s`** is for **S**pace.
-   **`w`** is for **W**ord.
-   **`b`** is for **B**oundary.

## The Golden Rule of Regex in Java

1.  If it's a **normal letter, number, or punctuation (like `,` or `:`)**, just type it: `","`
    
2.  If it is a **Regex Metacharacter (like `.`, `+`, `*`)**, escape it with two slashes: `"\\."`
    
3.  **Never** use a single slash unless you are trying to make a new line (`\n`) or a tab (`\t`).

# Character at i

s.getCharAt(i)  -- WRONG   
s.charAt(i) -- Correct Similar to length

## Immutability
String objects cannot be changed once created. Methods like s.charAt(i) only read characters; you cannot use them to modify the string, and expressions like s.charAt(i) = ... will **cause a compilation error**.

To bypass the same, work with character array or use stringbuilder's `setCharAt(index, character)`

## Performance Insight: Why cArr[i] over s.charAt(i)?
* Inside the loop, checking cArr[i] is significantly faster than calling s.charAt(i).
* s.charAt(i) is a method call that executes hidden index out-of-bounds checks every single time it runs.
* <mark class="note">Note</mark> cArr[i] is a direct primitive array lookup. It accesses the raw memory address instantly, making the loop lightweight and highly performant.

## why do we have s.charAt(i) and s.length() but not s.getCharAt(i) and s.getLength()? 

It comes down to a design choice made by the original creators of Java back in 1995, and it follows a specific naming convention standard called Getter/Setter Rules (specifically, the JavaBeans specification).

Here is the breakdown of why it is structured that way:

1. The Historical "Getter" Standard
In Java, standard "getter" methods usually look like getSomething(). For example, s.length() breaks this rule too (it doesn't have a get).

When Java was first built, the engineers decided that for core, heavily-used primitive types and data data structures (like Strings and Arrays), they wanted the method names to be as short, punchy, and explicit as possible.

2. Grammatical Precision
If you read s.charAt(i) out loud, it reads exactly like plain English:

"String, give me the character at index i."

Adding get would make it s.getCharAt(i), which sounds like "get character at index i". The creators felt charAt was concise, clear, and saved developers from typing an extra three letters millions of times a day.

3. Consistency across Languages
Java actually borrowed a lot of its initial style from C and C++. In older programming standards, abbreviations were highly prized to keep code dense. Keeping it as charAt felt familiar to the systems programmers migrating to Java at the time.

So, while it feels like it breaks the modern getX() convention you see everywhere else in Java objects, it’s just a legacy design choice that stuck because it's clean and easy to read out loud.

## getChars()

Copies characters from this string into the destination character array.

getChars(int srcBegin, int srcEnd, char[] dst, int dstBegin

The first character to be copied is at index srcBegin; the last character to be copied is at index srcEnd-1 (thus the total number of characters to be copied is srcEnd-srcBegin). The characters are copied into the subarray of dst starting at index dstBegin and ending at index

````java
String s = "Welcome";
System.out.println(s.length());
char[] ncr = new char[25];
System.out.println(ncr);
s.getChars(1,5,ncr,4);
System.out.println(ncr);
````
