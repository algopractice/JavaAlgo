```java
import java.util.HashSet;
import java.util.Set;

public class EmailValidator {
    public int numUniqueEmails(String[] emails) {
        Set<String> emailSet = new HashSet<>();

        for (String email : emails) {
            String[] arr = email.split("@");
            String localName = arr[0];
            String domain = arr[1];
            
            // Handle the '+' character rule
            String ignoredPlus = localName.split("\\+")[0];

            // Handle the '.' character rule
            // Better to use replace here
            StringBuilder sb = new StringBuilder();
            for (String str : ignoredPlus.split("\\.")) {
                sb.append(str);
            }
            
            emailSet.add(sb.append("@").append(domain).toString());
        }

        return emailSet.size();
    }
}
```

```java
// 1. Define your array of multiple inputs here
String[] inputEmails = {
    "test.email+alex@leetcode.com",
    "test.e.mail+bob.cathy@leetcode.com",
    "testemail+david@leetcode.com",
    "testemail.bob+cathy@leetcode.com",
    "testemail.cat+cathy@leetcode.com",
    "test@email.com"
};

// 2. Instantiate your class
EmailValidator validator = new EmailValidator();

// 3. Execute the method and print results
int uniqueCount = validator.numUniqueEmails(inputEmails);

System.out.println("Input Array: " + java.util.Arrays.toString(inputEmails));
System.out.println("Result     : " + uniqueCount + " unique email(s) found.");
```

# Cleaner

* **.replace()**
     * Instead of manually splitting by dots and looping, you can use .replace() to clean the local name efficiently in one or two lines.

```java
import java.util.HashSet;
import java.util.Set;

class Scratch {
    public static void main(String[] args) {
        String[] emails = {"test.email+alex@leetcode.com", "test.email@leetcode.com"};
        Set<String> emailSet = new HashSet<>();

        for (String email : emails) {
            String[] parts = email.split("@");
            String localName = parts[0];
            String domain = parts[1];

            // 1. Remove everything after the first '+'
            String cleanLocal = localName.split("\\+")[0];
            
            // 2. Remove all '.'
            cleanLocal = cleanLocal.replace(".", "");

            // 3. Combine and add to set
            emailSet.add(cleanLocal + "@" + domain);
        }

        System.out.println("Unique emails count: " + emailSet.size());
    }
}
```
