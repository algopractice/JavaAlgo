# Set

## Does `Set` have a `get()` method?

**No.** A `Set` in Java is designed to hold unique items and does not maintain a specific index-based order. Because there is no "index," there is no `get(int index)` method.

-   If you need to retrieve an item, you generally use an `Iterator` or a `for-each` loop.
    
-   For this problem, you don't actually need to "get" anything from the set; you only need to `add()` items to it and check its `size()` at the end.

* Set uses .add() and * not put()*. Put is for Map
* Also, you can't pass multiple individual characters into .add() all at once like that.
* Both are **Wrong**
    * vowels.add('a', 'e', 'i', 'o', 'u');
    * vowels.addAll('a', 'e', 'i', 'o', 'u');

Use `HashSet` when you need to store unique items and check for existence efficiently (the `add` and `contains` operations are $O(1)$ on average).
