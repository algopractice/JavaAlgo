# Entry

In Java, **`Map.Entry`** is an **interface nested within** the `Map` interface that represents a single **key-value pair** within a map.

When you iterate over a map, you are not just iterating over the keys or the values separately; you are iterating over the "entries" that bind them together.

### Why do we use it?

Using `Map.Entry` is the most efficient and standard way to access both the key and the value simultaneously during iteration.

-   **Performance:** If you iterate only using keys (e.g., `map.keySet()`) and then call `map.get(key)` inside the loop, you are essentially performing two lookups for every element. Accessing the `Entry` directly gives you the key and value in a single step, which is more efficient.
    
-   **Logical Coupling:** It reinforces the fact that the key and value are stored as a single unit in the underlying data structure (like a bucket in a `HashMap`).

### What is the "Entry" you saw?

You likely saw `Map.Entry` used frequently because it is the return type of the `entrySet()` method.

-   `map.entrySet()` returns a `Set` view of the mappings contained in the map.
    
-   Each element in that set is a `Map.Entry` object.
    
When you write `for (Map.Entry<Integer, Integer> e : fm.entrySet())`, you are telling Java to pull each of those individual key-value pairs out of the set one by one so you can inspect both parts of the pair without needing to look them up again.

## Raw type of Entry

**Must use Generics** with Entry, or cast the Object class returned by raw entry to the corresponding type

```java
import java.util.*;
```

```java
Map<Integer,Integer> m = new HashMap<>();
m.put(1,10);
m.put(2,20);
m.put(3,30);
m.put(4,40);
```

```java
int r = -1;
for (Map.Entry e : m.entrySet()) {
            if (e.getValue() == 30) {
                r = e.getKey();
                System.out.println("Key Found");

            } else {
                System.out.println("Not Found");
            }
        }
```

```java
int r = -1;
for (Map.Entry<Integer,Integer> e : m.entrySet()) {
            if (e.getValue() == 30) {
                r = e.getKey();
                System.out.println("Key Found");

            } else {
                System.out.println("Not Found");
            }
        }
```
